```python

import random
ucla_words = ['bruin','ucla','westwood','sproul','rieber', 'royce','powell','ackerman','kerckhoff','library','pauley','basketball','janss','bear','wooden','cssi','olympic','deneve','starship']
cs30_words = ['syntax','functions','debugging','list','string','boolean','dictionary','tuple','conditional','loop','turtle','map', 'filter','lambda','reduce','recursion','return','true','false']
LA_words = ['angeles','hollywood','rams','downtown','lakers','griffith','ucla','california', 'beverly','grove','socal','lax','koreatown','usc', 'getty','dodgers','universal']
def game():
  lives = 6
  choose = 1
  while choose == 1:
      version = input("Which version would you like to play?(uc/cs/la): ").lower()
      if version == 'uc':
        choose = 0
        chosen = ucla_words[(random.randint(1,len(ucla_words)-1))]
      elif version == 'cs':
        choose = 0
        chosen = cs30_words[(random.randint(1,len(cs30_words)-1))]
      elif version == 'la':
        choose = 0
        chosen = LA_words[(random.randint(1,len(LA_words)-1))]
      else:
        print ('Invalid')
  print ('The word is', len(chosen) ,'letters long')
  while lives > 0:
    place =[]
    letter = []
    none = []
    hp = []
    hl = []
    hidden = hp,hl
    guess = input('Please enter a guess: ').lower()
    if guess == chosen:
      print ('You won! Great game :)')
      lives = -1
      play_again()
      break
    if len(guess) != len(chosen):
      print ('Invalid')
      continue
    for i in range(len(chosen)):
      if guess[i] == chosen[i]:
        if guess.count(guess[i]) >= 2 and hidden.count(guess[i]) < chosen.count(guess[i]):
          if place.count(guess[i]) > 0:
            hp.append(guess[i])
            place.append(guess[i])
            place.append(place[(place.index(guess[i])+1)] + ', ' + str(i + 1))
            place.pop(place.index(guess[i])+1)
            place.remove(guess[i])
            if hp.count(guess[i]) >= chosen.count(guess[i]) and guess[i] in hl:
              hl.remove(guess[i])
              letter.pop(letter.index(guess[i])+1)
              letter.remove(guess[i])

          else:
            hp.append(guess[i])
            place.append(guess[i])
            place.append(' in position: '+ str(i+1))
        elif guess.count(guess[i]) == 1:
          hp.append(guess[i])
          place.append(guess[i])
        elif guess.count(guess[i]) > 1 and hidden.count(guess[i]) > chosen.count(guess[i]):
          if hp.count(guess[i]) > 0:
            hp.append(guess[i])
            place.append(guess[i])
            place.append(place[(place.index(guess[i])+1)] + ', ' + str(i + 1))
            place.pop(place.index(guess[i])+1)
            place.remove(guess[i])
            if hp.count(guess[i]) >= chosen.count(guess[i]) and guess[i] in hl:
              hl.remove(guess[i])
              letter.pop(letter.index(guess[i])+1)
              letter.remove(guess[i])
            if hp.count(guess[i]) >= chosen.count(guess[i]) and guess[i] in hl:
              hl.remove(guess[i])
              letter.pop(letter.index(guess[i])+1)
              letter.remove(guess[i])
          else:
            hp.append(guess[i])
            place.append(guess[i])
            place.append(' in position: '+ str(i+1))
            if guess[i] in hl:
              letter.pop(letter.index(guess[i])+1)
              letter.remove(guess[i])
              hl.remove(guess[i])
          if hp.count(guess[i]) >= chosen.count(guess[i]) and guess[i] in hl:
            hl.remove(guess[i])
            letter.pop(letter.index(guess[i])+1)
            letter.remove(guess[i])
        else:
          continue
      elif guess[i] in chosen and guess[i] != chosen[i]:
        if hl.count(guess[i]) > 0 and guess.count(guess[i]) > 1 and (hl.count(guess[i])+hp.count(guess[i])) < chosen.count(guess[i]):
          hl.append(guess[i])
          letter.append(guess[i])
          letter.append(letter[(letter.index(guess[i])+1)]+', ' + str(i + 1))
          letter.pop(letter.index(guess[i])+1)
          letter.remove(guess[i])
        if guess.count(guess[i]) >= 2 and (hl.count(guess[i])+hp.count(guess[i])) < chosen.count(guess[i]):
          hl.append(guess[i])
          letter.append(guess[i])
          letter.append(' in position: ' + str(i+1))
          if place.count(guess[i]) == chosen.count(guess[i]) and guess[i] in hl:
            letter.pop(letter.index(guess[i])+1)
            letter.remove(guess[i])
            hl.remove(guess[i])
        elif guess.count(guess[i]) == 1:
          hl.append(guess[i])
          letter.append(guess[i])
        if hp.count(guess[i]) > chosen.count(guess[i]) and guess[i] in hl:
          hl.remove(guess[i])
          letter.pop(letter.index(guess[i])+1)
          letter.remove(guess[i])
        if letter.count(guess[i]) > chosen.count(guess[i]):
          letter.pop(letter.index(guess[i])+1)
          letter.remove(guess[i])
          hl.remove(guess[i])
      elif guess[i] not in chosen:
        none.append(guess[i])
    for l in range(len(chosen)):
      if hp.count(guess[l]) >= chosen.count(guess[l]) and guess[l] in letter:
        hl.remove(guess[l])
        letter.pop(letter.index(guess[l])+1)
        letter.remove(guess[l])
    lives -= 1
    for i in none:
      if none.count(i) > 1:
        for j in range(none.count(i)-1):
          none.remove(i)
    if lives != -2:
      print('the following letter(s) are in the right position: ' + str(place))
      print ('the following letter(s) are in the wrong position: ' + str(letter))
      print ('the following letter(s) are not in the word: ' + str(none))
    if lives == 0:
      print ('You lost all your lives! The word was: ' + chosen + ". Better luck next time :)")
      play_again()
def play_again():
  retry = input("Would you like to play again?(y/n): ").lower()
  if retry == 'y':
    return game()
  elif retry == 'n':
    print('Thanks for playing!')
    pass
  else:
    print ('Invalid')
print ('Welcome to BRUINDLE, a Wordle alternative')
print ('You may choose to play between three versions: UCLA, CS30, or LA!')
yn = 1
while yn == 1:
  instructions = input("Would you like a brief overview of the game?(y/n): ").lower()
  if instructions == 'y':
    yn = 0
    print ('Alright, so the rules are simple. You will be given a random word and are instructed to guess what the word is. The length of the word will be given to you before the game begins. You will have 6 attempts per word and be told which letters are incorrect, which are correct, and which are in the correct position. Good luck!')
  elif instructions == 'n':
    yn = 0
    print ('Here we go.')
  else:
    print ('Invalid')
game()
```
